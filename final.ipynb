```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
import plotly
import plotly.express as px

#  Version check
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Matplotlib version:", matplotlib.__version__)
print("Seaborn version:", sns.__version__)
print("Plotly version:", plotly.__version__)


```

```python
df = pd.read_csv("customers-100.csv")
df.head()
```

```python
# Summary statistics
df.describe()

# Check for missing values
df.isnull().sum()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['City'], bins=5, kde=True, color='skyblue')
plt.title('Customer city Distribution')
plt.xlabel('City')
plt.ylabel('Count')
plt.grid(True)
plt.show()
```

```python
pivot = df.pivot_table(index='Country', values='Customer Id', aggfunc='count')
pivot = pivot.reset_index()
pivot.columns = ['Country', 'CustomerCount']  # Rename columns for clarity
pivot.head()
```

```python
import plotly.express as px

fig = px.bar(pivot, x='Country', y=pivot.columns[1], title="Customers by Country")
fig.show()
```
